```python
target_names=dict(enumerate(iris.target_names))
```

```python
df['species']=df['target'].map(target_names)
```

```python
df.groupby('species')[features_columns].mean()
```

```python
sns.set_style("darkgrid")
```

```python
#pairplot
plt.figure(figsize=(10,10))
sns.pairplot(df[features_columns+['species']],hue='species',diag_kind="kde")
plt.suptitle("pairwise feature relationship")
plt.show()
```

```python
sns.set_style("ticks")
```

```python
#pairplot
sns.set_style("whitegrid")
plt.figure(figsize=(10,10))
sns.pairplot(df[features_columns+['species']],hue='species',diag_kind="hist")
plt.suptitle("pairwise feature relationship")
plt.show()
```

```python
# correlation heatmap
sns.set_style("darkgrid")
corr=df[features_columns].corr(numeric_only=True)
plt.figure(figsize=(10,10))
sns.heatmap(corr,annot=True,cmap='coolwarm',linewidth=0.5)
plt.show()
```

```python
# correlation heatmap
sns.set_style("whitegrid")
plt.figure(figsize=(10,10))
sns.countplot(data=df,x='species')
plt.xlabel("species")
plt.ylabel('counts')
plt.show()

# each classs is evenly distributed
```

```python
# correlation heatmap
sns.set_style("whitegrid")
plt.figure(figsize=(10,10))
sns.relplot(data=df,x="petal length (cm)",y="petal width (cm)",hue='species')
plt.title("petal length  vs petal width")
plt.show()
```
